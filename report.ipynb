Do you notice any mathematical relationship between f(d, 4) and k(d − 1)? State this
relationship, and briefly explain why it should be true for all values of d.

I notice a relationship of f(d, 4) = 2k(d − 1) + 1, for all values of d that are greater than or equal to 1.

The largest disk cannot move to the final peg until the other d - 1 disks are all cleared onto the two spare pegs, which takes at least k(d - 1) moves. Moving the largest disk takes 1 move (hence the + 1 at the end of the equation). You can then stack the d - 1 disks on top of it in another k(d - 1) moves by running that initial sequence in reverse with the start and target pegs swapped, thus giving 2k(d - 1) + 1 moves in total.

## 2.2 Implementing Information Gain

Train a decision tree on churn train.csv and print, in a readable text form, the first three
levels of the tree.

You should observe that a deep tree can achieve excellent training accuracy on churn data,
but may overfit, because local greedy choices can capture noise and small customer subgroups.
5

Run your experiment for maximum depths of 2, 3, 4, 5, and unrestricted depth. For each
setting, print:

- the maximum depth allowed (the setting) and the depth actually reached by the learned
tree,
- the number of internal nodes,
- training, validation and test accuracy, and the corresponding F1 scores.

State which depth you would choose and why.

```python
from pathlib import Path
from trees import (load_csv, DecisionTree, accuracy_score, f1_score_positive)
base = Path(".")
training_set = load_csv(base / "data" / "churn_train.csv")
validation_set = load_csv(base / "data" / "churn_valid.csv")
test_set  = load_csv(base / "data" / "churn_test.csv")
depth_settings = [2, 3, 4, 5, None]

def scores(model, rows):
    y = []
    for r in rows:
        y.append(r["churned"])
    
    p = model.predict(rows)
    return accuracy_score(y, p), f1_score_positive(y, p)

for setting in depth_settings:
    tree = DecisionTree(criterion="information_gain", max_depth=setting)
    tree.fit(training_set)

    internal_nodes, leaves = tree.count_nodes()
    train_acc, train_f1 = scores(tree, training_set)
    valid_acc, valid_f1 = scores(tree, validation_set)
    test_acc, test_f1 = scores(tree, test_set)

    label = "unrestricted" if setting is None else setting
    print(f"Max depth: {label} - Depth reached: {tree.tree_depth()}")
    print(f"Internal nodes: {internal_nodes}")
    print(f"Train: accuracy={train_acc:.4f}, F1={train_f1:.4f}")
    print(f"Valid: accuracy={valid_acc:.4f}, F1={valid_f1:.4f}")
    print(f"Test:  accuracy={test_acc:.4f}, F1={test_f1:.4f}\n")
```

I would chose a depth of 3 nodes. This is due to it's testing accuracy being the highest (72%). Depths greater than three have higher training scores but overfit to the data and thus have lower testing scores. Depths lower than three underfit to the data and have low accuracy scores throughout.

Numerical splitting is more computational expensive because for each numeric attribute the evaluation function must search over many candiate thresholds, with each candidate requiring a full scan of the data. If there are 'm' distinct values, the time complexity to train the model becomes O(n x m), with the worst case scenario scaling up to O(n^2). Categorical splitting only needs one iteration of the data set to group rows into category and calculate information gain, making thus having a time complexity of O(n).

## 2.4.4 Early Stopping Rule

Design your own pruning or early-stopping rule for entropy-based decision trees and implement
it in both your bagging and random forest models. For both methods, with and without your
rule, record the training time and the training, validation and test accuracy.

Explain your rule, justify its design, and discuss its effect on the training process. Finally,
report whether your rule improves or degrades validation and test accuracy and F1 score.

```python
import time

# Early-stopping rule: do not split if the best information gain <= min_gain.
num_trees = 20
MIN_GAINS = (0.0, 0.005, 0.01, 0.02, 0.05)

configs = [
    ("Bagging", None),
    (f"Random forest (m = {sqrt_feature_subsample_size})", sqrt_feature_subsample_size),
]


def report_min_gain_result(name, min_gain, model, train_time):
    label = "with rule" if min_gain > 0 else "without rule"
    train_accuracy, train_f1 = scores(model, training_set)
    validation_accuracy, validation_f1 = scores(model, validation_set)
    test_accuracy, test_f1 = scores(model, test_set)
    print(f"{name}, {label} (min_gain = {min_gain})")
    print(f"Training time: {train_time:.2f}s")
    print(f"Average depth: {model.average_depth():.2f}")
    print(f"Train: accuracy={train_accuracy:.4f}, F1={train_f1:.4f}")
    print(f"Valid: accuracy={validation_accuracy:.4f}, F1={validation_f1:.4f}")
    print(f"Test: accuracy={test_accuracy:.4f}, F1={test_f1:.4f}\n")


print(f"Total number of trees: {num_trees}\n")

for name, feature_subsample in configs:
    for min_gain in MIN_GAINS:
        model = BaggingEnsemble(
            num_trees=num_trees,
            criterion="information_gain",
            feature_subsample_size=feature_subsample,
            random_state=3411,
            min_gain=min_gain,
        )
        start = time.perf_counter()
        model.fit(training_set)
        train_time = time.perf_counter() - start
        report_min_gain_result(name, min_gain, model, train_time)
```

The rule (at it's varying levels) significantly improves training time. However, this comes at a cost of testing accuracy degradation, especially when minimum information gain increases. With higher levels of information gain (0.05) F1 scores plummet to 0 (for random forests only) because trees stop splitting almost immediately because most splits are not above the 0.05 information gain threshold. 

Small thresholds such as 0.01 and 0.02 have no negative effect to the testing and F1 accuracy (for bagging), but also have marginal training time improvements with their non-rule counterparts. This rule is best used to observe general model accuracy over large amounts of data. The data being used in the question is only around 1,000 entries, and therefore, regardless of whether an information gain rule is being applied, both versions take a few seconds to apply. However as the data set increases in size, training time increases significantly and thus, it can be advantageous to implement a `min_gain` rule to quickly gauge model accuracy.

The rule does not clearly improve testing or validation accuracy, and trades quality for speed.
