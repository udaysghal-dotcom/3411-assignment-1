## 2.2 Implementing Information Gain

Train a decision tree on churn train.csv and print, in a readable text form, the first three
levels of the tree.

You should observe that a deep tree can achieve excellent training accuracy on churn data,
but may overfit, because local greedy choices can capture noise and small customer subgroups.
5

Run your experiment for maximum depths of 2, 3, 4, 5, and unrestricted depth. For each
setting, print:

- the maximum depth allowed (the setting) and the depth actually reached by the learned
tree,
- the number of internal nodes,
- training, validation and test accuracy, and the corresponding F1 scores.

State which depth you would choose and why.

```python
from pathlib import Path
from trees import (load_csv, DecisionTree, accuracy_score, f1_score_positive)
base = Path(".")
training_set = load_csv(base / "data" / "churn_train.csv")
validation_set = load_csv(base / "data" / "churn_valid.csv")
test_set  = load_csv(base / "data" / "churn_test.csv")

depth_settings = [2, 3, 4, 5, None]

def scores(model, rows):
    y = []
    for r in rows:
        y.append(r["churned"])
    
    p = model.predict(rows)
    return accuracy_score(y, p), f1_score_positive(y, p)

for setting in depth_settings:
    tree = DecisionTree(criterion="information_gain", max_depth=setting)
    tree.fit(training_set)

    internal_nodes, leaves = tree.count_nodes()
    train_acc, train_f1 = scores(tree, training_set)
    valid_acc, valid_f1 = scores(tree, validation_set)
    test_acc, test_f1 = scores(tree, test_set)

    label = "unrestricted" if setting is None else setting
    print(f"Max depth: {label} - Depth reached: {tree.tree_depth()}")
    print(f"Internal nodes: {internal_nodes}")
    print(f"Train: accuracy={train_acc:.4f}, F1={train_f1:.4f}")
    print(f"Valid: accuracy={valid_acc:.4f}, F1={valid_f1:.4f}")
    print(f"Test:  accuracy={test_acc:.4f}, F1={test_f1:.4f}")
    print()
```

I would chose a depth of 3 nodes. This is due to it's testing accuracy being the highest (72%). Depths greater than three have higher training scores but overfit to the data and thus have lower testing scores. Depths lower than three underfit to the data and have low accuracy scores throughout.

## 2.3 Numerical Threshold Search

Print the best threshold chosen for each of the following attributes when they are evaluated
at the root node: tenure months, monthly charges, avg monthly gb, and support tickets 90d.

Briefly discuss why numerical threshold search is more computationally expensive than
categorical splitting. Looking at your learned trees, what types of customer behaviour seem
most predictive of churn?

```python
from pathlib import Path
from trees import load_csv, DecisionTree

training_set = load_csv(Path(".") / "data" / "churn_train.csv")

features = [
    "tenure_months",
    "monthly_charges",
    "avg_monthly_gb",
    "support_tickets_90d",
]

tree = DecisionTree(criterion="information_gain", random_state=3411)
for feature in features:
    split = tree.best_numeric_split(training_set, feature)
    print(f"{feature}: threshold = {split.threshold:.3f}, information gain = {split.score:.4f}")

print()
print("First three levels, with numeric and categorical features")
tree.fit(training_set)
tree.print_tree(max_depth=2)
```
